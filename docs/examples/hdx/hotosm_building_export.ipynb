# HOTOSM OpenStreetMap export (buildings)

Resolve a HOTOSM country building export and list its resources, then download the GeoPackage. **Live** public query; wrapped for offline `nbval-lax`. HOTOSM exports can be large — Uruguay is used here as a smaller example.

```python
from pathlib import Path

from earthlens import EarthLens

OUT_DIR = Path('hdx_output')
OUT_DIR.mkdir(exist_ok=True)
```

```python
# Inspect the resources without downloading (cheap metadata call).
try:
    backend = EarthLens(
        data_source="hdx",
        dataset='hotosm-uruguay-buildings',
        variables=['Geopackage'],
        path=str(OUT_DIR),
    )
    products = backend._search()
    for p in products:
        print(p.metadata['name'], '|', p.metadata['format'])
except Exception as exc:
    print(f'skipped search: {type(exc).__name__}: {exc}')
```

```python
paths = None
try:
    paths = EarthLens(
        data_source='hdx',
        dataset='hotosm-uruguay-buildings',
        variables=['Geopackage'],
        path=str(OUT_DIR),
    ).download(progress_bar=False)
    print('downloaded:', [Path(p).name for p in paths])
except Exception as exc:
    print(f'skipped live query: {type(exc).__name__}: {exc}')
```
